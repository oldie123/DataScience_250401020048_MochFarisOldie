Notebook ini membahas struktur data Python, operasi dasar NumPy, pembuatan Series/DataFrame Pandas, dan eksplorasi awal dataset Titanic.

```python
import pandas as pd
import numpy as np

# A. Struktur Data Python
data = [10, 20, "Python", True, 3.14]
print("List:", data)
print("Elemen pertama:", data[0])
print("Elemen terakhir:", data[-1])
print("Slicing 1:3:", data[1:3])

data.append(99)
data.insert(1, 55)
data.remove(20)
print("Setelah manipulasi:", data)

mahasiswa = {"nama": "Moch Faris Oldie", "nim": "250401020048", "prodi": "Informatika", "ipk": 3.85}
print()
print("Dictionary:", mahasiswa)
print("Nama:", mahasiswa["nama"])

himpunan = {1, 2, 2, 3, 4, 4, 5}
print()
print("Set:", himpunan)

# B. NumPy
arr = np.array([1, 2, 3, 4, 5])
print()
print("Array NumPy:", arr)
print("Mean:", arr.mean())
print("Std:", arr.std())
print("Array + 10:", arr + 10)

mat = np.array([[1, 2], [3, 4], [5, 6]])
print("Shape matriks:", mat.shape)
print("Transpose:")
print(mat.T)

# C. Pandas dasar
df_series = pd.Series([85, 92, 78, 95, 88], index=["Alice", "Bob", "Caca", "Dina", "Eko"])
print()
print("Series:")
print(df_series)
print("Mean Series:", df_series.mean())

df = pd.DataFrame({"nama": ["Ali", "Budi", "Caca", "Dina"], "usia": [22, 25, 21, 23], "kota": ["Jakarta", "Bandung", "Surabaya", "Medan"], "ipk": [3.8, 3.5, 3.9, 3.7]})
print()
print("DataFrame:")
print(df)
print("Shape:", df.shape)

# D. Titanic Exercises
def load_titanic():
    url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
    try:
        return pd.read_csv(url)
    except Exception as error:
        print("Gagal mengambil dataset online, memakai data contoh lokal:", error)
        return pd.DataFrame({
            "PassengerId": range(1, 11),
            "Survived": [0, 1, 1, 1, 0, 0, 1, 0, 1, 0],
            "Pclass": [3, 1, 3, 1, 3, 2, 1, 3, 2, 3],
            "Name": [f"Passenger {i}" for i in range(1, 11)],
            "Sex": ["male", "female", "female", "female", "male", "male", "female", "male", "female", "male"],
            "Age": [22, 38, 26, 35, 35, np.nan, 54, 2, 27, np.nan],
            "SibSp": [1, 1, 0, 1, 0, 0, 1, 3, 0, 0],
            "Parch": [0, 0, 0, 0, 0, 0, 0, 1, 2, 0],
            "Ticket": ["A/5", "PC", "STON", "113803", "373450", "330877", "17463", "349909", "347742", "237736"],
            "Fare": [7.25, 71.28, 7.92, 53.1, 8.05, 8.46, 51.86, 21.08, 11.13, 30.07],
            "Cabin": [np.nan, "C85", np.nan, "C123", np.nan, np.nan, "E46", np.nan, np.nan, np.nan],
            "Embarked": ["S", "C", "S", "S", "S", "Q", "S", "S", "S", np.nan],
        })

titanic = load_titanic()
print()
print("Q1")
print("Jumlah baris dan kolom:", titanic.shape)
print("Kolom yang tersedia:")
print(titanic.columns.tolist())

print()
print("Q2")
missing_count = titanic.isnull().sum()
missing_pct = (titanic.isnull().mean() * 100).round(2)
missing_table = pd.DataFrame({"missing_count": missing_count, "missing_pct": missing_pct})
print(missing_table[missing_table["missing_count"] > 0])

print()
print("Q3")
survival_rate = titanic["Survived"].mean() * 100
print(f"Persentase penumpang selamat: {survival_rate:.2f}%")

print()
print("Q4")
women_class1 = titanic[(titanic["Sex"] == "female") & (titanic["Pclass"] == 1)]
print("Jumlah wanita kelas 1:", len(women_class1))
print("Rata-rata usia:", women_class1["Age"].mean())

print()
print("Q5")
survival_by_class = titanic.groupby("Pclass")["Survived"].mean().sort_values(ascending=False)
print(survival_by_class)
print("Kelas dengan survival rate tertinggi:", survival_by_class.idxmax())

print()
print("Q6")
analysis_by_sex = titanic.groupby("Sex")["Survived"].mean().sort_values(ascending=False)
print("Survival rate berdasarkan gender:")
print(analysis_by_sex)
analysis_by_embarked = titanic.groupby("Embarked")["Survived"].mean().sort_values(ascending=False)
print("Survival rate berdasarkan pelabuhan keberangkatan:")
print(analysis_by_embarked)
```
